```python
import sys
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))

from src.data.load_data import load_ratings
from src.models.nmf import (
    prepare_surprise_data,
    train_nmf_model,
)

DATA_DIR = PROJECT_ROOT / "data" / "raw"
```

```python
ratings = load_ratings(DATA_DIR)

ratings.head()
```

```python
data = prepare_surprise_data(
    ratings,
    rating_scale=(1, 5),
)
```

```python
from surprise.model_selection import train_test_split

trainset, testset = train_test_split(
    data,
    test_size=0.20,
    random_state=42,
)
```

```python
model_nmf = train_nmf_model(
    trainset=trainset,
    n_factors=15,
    random_state=42,
)
```

```python
from surprise import accuracy

predictions_nmf = model_nmf.test(testset)

nmf_rmse = accuracy.rmse(
    predictions_nmf,
)
```
